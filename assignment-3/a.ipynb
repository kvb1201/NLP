>- Starts with an lowercase letter
>- Followed by zero or more lowercase letters

design a finite state transducer (FST) to generate the morph/grammatical features for
every word in the corpus. 

>Your output should look like the following:
>- foxes = fox+N+PL (can be generalized as root+category+number)
>- fox = fox+N+SG [where SG->Singular, PL->Plural, N->Noun]
>- You need to take the following properties:
>- Name Rule Description Example
E insertion e is added after -s, -z, -x,
-ch, -sh before -s is added

watch/watches, fox/foxes

>- Y replacement -y changes to -ie before -s try/tries
>- S addition -s is added at the end bag/bags
You need to ensure that incorrect words are not generated. You return an output “Invalid
Word” in that case. Example: foxs = “Invalid Word”

```python
from pathlib import Path
path = Path('brown_nouns.txt')
data = []
with path.open('r', encoding="utf") as f:
    for i,word in enumerate(f):
        if i<10:
            print(word)
        data.append(word.strip())
    
```

```python
len(data)
word_dict = set(data)
```

### Finite State Tranducer which performs morphological analysis.
assume the transitions which do not have letters are all letters except the one with transition.
![dfa image](fst_3.png)

```python
class fst:
    def __init__(self, start_state, input_symbols, output_symbols, state_transition, output_function, final_state):
        self.start_state = start_state
        self.input_symbols = input_symbols
        self.output_symbols = output_symbols
        self.state_transition = state_transition
        self.output_function = output_function
        self.final_state = final_state

    def correct(self, string:str):
        end = string.find("+")
        base_word = string[:end]

        if string.find('+SG')!=-1 or string.find('+invalid')!=-1:
            return string if base_word in word_dict else "invalid word"
        i_1 = base_word.rfind("ies")
        if i_1 != -1:
            candidate = base_word[:i_1] +'y'
            if candidate in word_dict:
                return candidate + string[end:]
            else:
                return base_word + "+N+SG" if base_word in word_dict else "invalid word"
        i_2 = base_word.rfind('es')
        if i_2 != -1:
            candidate = base_word[:i_2]
            if candidate in word_dict:
                return base_word[:i_2] + string[end:]
            else:
                return base_word + "+N+SG" if base_word in word_dict else "invalid word"
        n = len(base_word)
        if n >1:
            candidate = base_word[:n-1]
            if candidate in word_dict:
                return base_word[:n-1] + string[end:]
            else:
                return base_word + "+N+SG" if base_word in word_dict else "invalid word"
        return "invalid word"

        
    def parse(self, string):
        curr_state = self.start_state
        #Special fst specifically for this problem: Consists of one final state: qend which is only reached when the string is completed
        qend = self.final_state
        analysis = []
        i =0
        while curr_state != qend:
            curr_input = "" if i >= len(string) else string[i]
            i+=1
            new_state = self.state_transition[curr_state][curr_input]
            curr_output = self.output_function[curr_state][curr_input]
            curr_state = new_state
            if curr_output != "":
                analysis.append(curr_output)

        analysed = "".join(analysis)
        final = self.correct(analysed)
        return final 

        
        

```

### Transitions and fst defintion

>
>- normal character → qst
>- c                → qc
>- x,z              → q_special
>- s                → q_s
>- EOF              → qend (+N+SG)

```python
start_state = 'qst'
final_state = 'qend'

states = [
    'qst',
    'qc',
    'q_special',
    'qe',
    'qes',
    'q_s_error',
    'q_s',
    'qend'
]
```

```python
import string

inputs = list(string.ascii_letters)

for word in data:
    for ch in word.strip():
        if ch not in inputs:
            inputs.append(ch)

inputs.append('')

outputs = list(string.ascii_letters) + [
    '', '+N+SG', '+N+PL', '+invalid'
]

# States
start_state = 'qst'
final_state = 'qend'

states = [
    'qst',
    'qc',
    'q_special',
    'qe',
    'qes',
    'q_s_error',
    'q_s',
    'qend'
]


# qst

qst_transitions = {}
qst_output = {}

for ch in inputs:

    if ch not in ['c', 's', 'x', 'z', '']:
        qst_transitions[ch] = 'qst'
        qst_output[ch] = ch

qst_transitions['c'] = 'qc'
qst_output['c'] = 'c'

qst_transitions['s'] = 'q_s'
qst_output['s'] = 's'

for ch in ['x', 'z']:
    qst_transitions[ch] = 'q_special'
    qst_output[ch] = ch

qst_transitions[''] = 'qend'
qst_output[''] = '+N+SG'


# qc

qc_transitions = {}
qc_output = {}

for ch in inputs:

    if ch == 'h':
        qc_transitions[ch] = 'q_special'
        qc_output[ch] = 'h'

    elif ch == '':
        qc_transitions[ch] = 'qend'
        qc_output[ch] = '+N+SG'

    else:
        qc_transitions[ch] = 'qst'
        qc_output[ch] = ch


# q_special

q_special_transitions = {}
q_special_output = {}

for ch in inputs:

    if ch == 'e':
        q_special_transitions[ch] = 'qe'
        q_special_output[ch] = 'e'

    elif ch == 's':
        q_special_transitions[ch] = 'q_s_error'
        q_special_output[ch] = 's'

    elif ch == '':
        q_special_transitions[ch] = 'qend'
        q_special_output[ch] = '+N+SG'

    else:
        q_special_transitions[ch] = 'qst'
        q_special_output[ch] = ch


# q_s

q_s_transitions = {}
q_s_output = {}

for ch in inputs:

    if ch == '':
        q_s_transitions[ch] = 'qend'
        q_s_output[ch] = '+N+PL'

    elif ch == 'e':
        q_s_transitions[ch] = 'qe'
        q_s_output[ch] = 'e'

    elif ch == 'h':
        q_s_transitions[ch] = 'q_special'
        q_s_output[ch] = 'h'

    else:
        q_s_transitions[ch] = 'qst'
        q_s_output[ch] = ch


# qe

qe_transitions = {}
qe_output = {}

for ch in inputs:

    if ch == 's':
        qe_transitions[ch] = 'qes'
        qe_output[ch] = 's'

    elif ch == '':
        qe_transitions[ch] = 'qend'
        qe_output[ch] = '+N+SG'

    else:
        qe_transitions[ch] = 'qst'
        qe_output[ch] = ch


# qes

qes_transitions = {}
qes_output = {}

for ch in inputs:

    if ch == '':
        qes_transitions[ch] = 'qend'
        qes_output[ch] = '+N+PL'

    else:
        qes_transitions[ch] = 'qst'
        qes_output[ch] = ch


# q_s_error

q_s_error_transitions = {}
q_s_error_output = {}

for ch in inputs:

    if ch == '':
        q_s_error_transitions[ch] = 'qend'
        q_s_error_output[ch] = '+invalid'

    else:
        q_s_error_transitions[ch] = 'qst'
        q_s_error_output[ch] = ch


transitions = {
    'qst': qst_transitions,
    'qc': qc_transitions,
    'q_special': q_special_transitions,
    'qe': qe_transitions,
    'qes': qes_transitions,
    'q_s': q_s_transitions,
    'q_s_error': q_s_error_transitions
}

output_function = {
    'qst': qst_output,
    'qc': qc_output,
    'q_special': q_special_output,
    'qe': qe_output,
    'qes': qes_output,
    'q_s': q_s_output,
    'q_s_error': q_s_error_output
}
```

```python
my_fst = fst(
    start_state=start_state,
    input_symbols=inputs,
    output_symbols=outputs,
    state_transition=transitions,
    output_function=output_function,
    final_state=final_state
)
```
