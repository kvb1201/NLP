### Tokenization Strategy
Single regex based sentence tokenizer fails in detecting boundaries around special strings such as decimals, abbreviations, etc.
The below strategy predetects the special string boundaries and skips occurences of end symbols in those strings

```python
def sentence_tokenizer(text):
    spans = [(m.start(), m.end()) for m in re.finditer(special,text)]

    sentence = []
    start =0
    span_idx =0

    for m in end.finditer(text):
        pos = m.start()

        while span_idx <len(spans) and spans[span_idx][1] <= pos:
            span_idx+=1

        if span_idx < len(spans) and spans[span_idx][0] <= pos < spans[span_idx][1]:
            continue

        sentence.append(text[start:pos+1].strip())
        start = pos+1

    if start < len(text):
        sentence.append(text[start:].strip())

    return [s for s in sentence if s]
```

#### Word Tokenization Issues
The issue arises when we want to tokenization punctuations separately. Handling punctuations requires that the punctuation does not occur in special strings. Since the sentence token strategy is based on defining boundaries on appearance symbol, we may split the last word as word + punctuation.

```python

def word_tokenizer(sentence):
    words = re.findall(word, sentence)
    last_word = words.pop()
    word_len = len(last_word)
    if word_len >=2:
        words.append(last_word[:word_len-1])
        words.append(last_word[word_len-1:])
    return words

print(word_tokenizer(test_1[0]))
        

```

```python
output1 = Path("tokenized_sentences.txt")
output2 = Path("tokenized_words.txt")
sentence_count =0
lines =0
word_count =0
unique_words = set()
char_count =0
with gu_data.open('r', encoding='utf-8') as fin, output1.open('w', encoding = 'utf-8') as fout1,\
    output2.open('w', encoding = "utf-8") as fout2:
    for line in fin:
        lines+=1

        if(lines%10000 ==0):
            print(f"{lines} processed")
            fout1.flush()
            fout2.flush()

        for sentence in sentence_tokenizer(line):
            fout1.write(sentence + '\n')
            sentence_count +=1
            for w in word_tokenizer(sentence):
                unique_words.add(w)
                fout2.write(w +" ")
                word_count+=1
                char_count += len(w)
            fout2.write("\n")


print("Sentence Produced = ", sentence_count)
print("lines processed = ", lines)
print("Word Count =", word_count)
print("Character Count =", char_count)
type_token_ratio = len(unique_words)/word_count
avg_sentence_len = word_count/sentence_count
avg_word_len = char_count/word_count
with summary.open('a',encoding = 'utf-8') as s:
    s.write(f"Sentence tokenized = {sentence_count}\n")
    s.write(f"Words Produced = {word_count} \n")
    s.write(f"Character Count = {char_count}\n")
    s.write(f"Average Sentence Lengh = {avg_sentence_len}\n")
    s.write(f"Average word length = {avg_word_len}\n")
    s.write(f"Type/Token Ratio (TTR) = {type_token_ratio}\n")
    


```

```python
from pathlib import Path
import pandas as pd

def txt_to_parquet(txt_file, out_dir, chunk_size=100000):
    out_dir = Path(out_dir)
    out_dir.mkdir(exist_ok=True)

    chunk = []
    part = 0

    with open(txt_file, "r", encoding="utf-8") as fin:
        for line in fin:
            chunk.append(line.rstrip("\n"))

            if len(chunk) == chunk_size:
                df = pd.DataFrame({"sentence": chunk})

                df.to_parquet(
                    out_dir / f"part_{part:04d}.parquet",
                    index=False,
                    compression="snappy",
                )

                chunk.clear()
                part += 1

        # remaining rows
        if chunk:
            df = pd.DataFrame({"sentence": chunk})
            df.to_parquet(
                out_dir / f"part_{part:04d}.parquet",
                index=False,
                compression="snappy",
            )
```
